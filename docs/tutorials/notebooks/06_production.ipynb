# Production Tips

Best practices for production deployment.

## Agent lifetime and sessions

Use the asynchronous factory and close the agent with its context manager. A chat
session is created synchronously; sending a turn is asynchronous.

```python
from victor.framework import Agent

async with await Agent.create(provider="ollama", model="qwen2.5-coder:7b") as agent:
    result = await agent.run("Summarize the repository")
    print(result.content)
    session = agent.chat("Help me investigate the build")
    response = await session.send("Start with the project configuration")
    print(response.content)
```

`Agent.run` does not accept a `session_id` keyword. Use the session interface for
multi-turn conversation. Provider retries and fallback are configured through
runtime provider services; `RateLimiter` is not exported by `victor.framework`.
